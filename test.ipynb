```python
import numpy as np
from glob import glob
from matplotlib import pyplot as plt
import pypff
%matplotlib widget
```

```python
dirs = sorted(glob('obs_UCB*'))
#dir_index = -3
dir_index = -1
files = sorted(glob('%s/*_ph256*'%dirs[dir_index]))
print(dirs[dir_index])
print(files)
```

```python
ph256_dpff = pypff.io.datapff(files[0])
ph256_d, ph256_md = ph256_dpff.readpff( ver='qfp', metadata=True)
```

```python
fig = plt.figure()
subfig = fig.add_subplot(111)

subfig.plot(ph256_d[0])
subfig.set_xlabel('Pixels')
subfig.set_ylabel('Pulse Height')
subfig.set_title('PH data with 1 CTEST enabled', size=16)
```
